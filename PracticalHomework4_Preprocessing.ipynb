### ACS Data

Below, we import five different ACS datasets of demographic data for King County census tracts. These datasets are then adjusted to ensure each row represents a given census tract. They are then all merged together into a single DataFrame ``acs``.

```python
income = pd.read_csv('C:/Users/jacob/Downloads/B19013.csv')
tenure = pd.read_csv('C:/Users/jacob/Downloads/B25003.csv')
race = pd.read_csv('C:/Users/jacob/Downloads/B02001.csv')
home_value = pd.read_csv('C:/Users/jacob/Downloads/B25077.csv')
rent_burden = pd.read_csv('C:/Users/jacob/Downloads/B25070.csv')
```

```python
tenure.head()
```

```python
tenure = tenure.loc[:, ~tenure.columns.str.contains('Margin')]
```

```python
tenure = tenure.set_index("Label (Grouping)")
```

```python
tenure = tenure.T
```

```python
tenure.head()
```

```python
income = income.loc[:, ~income.columns.str.contains('Margin')]
```

```python
income = income.set_index('Label (Grouping)')
income = income.T
```

```python
income.head()
```

```python
race = race.loc[:, ~race.columns.str.contains('Margin')]
```

```python
race = race.set_index('Label (Grouping)')
race = race.T
```

```python
race.head()
```

```python
home_value = home_value.loc[:, ~home_value.columns.str.contains('Margin')]
```

```python
home_value = home_value.set_index('Label (Grouping)')
home_value = home_value.T
```

```python
home_value.head()
```

```python
rent_burden = rent_burden.loc[:, ~rent_burden.columns.str.contains('Margin')]
```

```python
rent_burden = rent_burden.set_index('Label (Grouping)')
rent_burden = rent_burden.T
```

```python
rent_burden.head()
```

```python
# set each index as new column
income = income.reset_index()
tenure = tenure.reset_index()
race = race.reset_index()
home_value = home_value.reset_index()
rent_burden = rent_burden.reset_index()
```

```python
income = income.rename(columns={income.columns[0]: 'tractID'})
tenure = tenure.rename(columns={tenure.columns[0]: 'tractID'})
race = race.rename(columns={race.columns[0]: 'tractID'})
home_value = home_value.rename(columns={home_value.columns[0]: 'tractID'})
rent_burden = rent_burden.rename(columns={rent_burden.columns[0]: 'tractID'})
```

```python
tenure.head()
```

Clean up tract ID 

```python
income['tractID'] = income['tractID'].str.extract(r'Census Tract ([0-9.]+)')
```

```python
income.head()
```

```python
tenure['tractID'] = tenure['tractID'].str.extract(r'Census Tract ([0-9.]+)')
race['tractID'] = race['tractID'].str.extract(r'Census Tract ([0-9.]+)')
home_value['tractID'] = home_value['tractID'].str.extract(r'Census Tract ([0-9.]+)')
rent_burden['tractID'] = rent_burden['tractID'].str.extract(r'Census Tract ([0-9.]+)')
```

```python
acs = income.copy()
```

```python
acs = acs.merge(tenure, on=['tractID'], how='inner')
acs = acs.merge(race, on=['tractID'], how='inner')
acs = acs.merge(home_value, on=['tractID'], how='inner')
acs = acs.merge(rent_burden, on=['tractID'], how='inner')
```
